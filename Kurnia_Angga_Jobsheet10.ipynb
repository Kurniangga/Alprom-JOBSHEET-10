Praktikum 01 – Bubble Sort

```python
def bubble_sort(data):
  n = len (data)
  for i in range (n):
      for j in range(0, n - i - 1):
         if data[j] > data[j + 1]:
            data[j], data[j + 1] = data[j + 1], data[j]
  return data

angka = [64, 34, 25, 12, 22, 11, 90]
print ("Data sebelum diurutkan: ", angka)
print ("Data setelah Bubble Sort: ", bubble_sort(angka))
```

Praktikum 02 – Selection Sort

```python
def selection_sort(data):
  n = len(data)
  for i in range(n):
    min_index = i
    for j in range(i + 1, n):
        if data[j] < data[min_index]: min_index = j
    data[i], data[min_index] = data[min_index], data[i]
  return data

angka = [64, 25, 12, 22, 11]
print("Data sebelum diurutkan:", angka)
print("Data setelah Selection Sort:", selection_sort(angka))
```

Praktikum 03 – Insertion Sort

```python
def insertion_sort(data):
  n = len(data)
  for i in range(1, n):
        key = data[i]
        j = i -1
        while j >= 0 and data[j] > key:
            data[j + 1] = data[j]
            j -= 1
        data[j+1] = key
  return data

angka = [12, 11, 13, 5, 6]
print ("Data sebelum diurutkan: ", angka)
print ("Data setelah Insertion Sort: ", insertion_sort(angka))

```

Praktikum 04 – Studi Kasus: Mengurutkan Nilai Mahasiswa

```python
def bubble_sort_nilai (data):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if data[j][1] < data[j + 1][1]:
        data[j], data[j + 1] = data[j + 1], data[j]
  return data

mahasiswa = [("Andi", 80), ("Budi", 70), ("Cici", 90), ("Doni", 85)]
print ("Data sebelum diurutkan: ", mahasiswa)
print ("Data setelah diurutkan (desc):", bubble_sort_nilai(mahasiswa))
```

Praktikum 06 – Sorting Data Barang (Nama & Harga)

```python
def bubble_sort_key(data, key_index=0, reverse=False):
  n = len(data)
  for i in range (n):
    for j in range(0, n - i - 1):
      if reverse:
        if data[j][key_index] < data[j + 1][key_index]:
          data[j], data[j + 1] = data[j + 1], data[j]
      else:
        if data[j][key_index] > data[j + 1][key_index]:
          data[j], data[j + 1] = data[j + 1], data[j]
  return data

barang = [("Buku", 15000), ("Pensil", 5000), ("Tas", 75000), ("Pulpen", 10000)]
print("Urut berdasarkan nama (asc):", bubble_sort_key(barang.copy(), 0))
print("Urut berdasarkan harga (desc):", bubble_sort_key(barang.copy(), 1, reverse=True))
```

Penugasan No 3 - Eksperimen Bubble Sort

```python
def bubble_sort_descending(data):
    n = len(data)
    for i in range(n):
        for j in range(0, n - i - 1):
            if data[j] < data[j + 1]:
                data[j], data[j + 1] = data[j + 1], data[j]
    return data

angka = [64, 34, 25, 12, 22, 11, 90]

print("Data sebelum diurutkan: ", angka)
print("Data setelah Bubble Sort (Descending): ", bubble_sort_descending(angka))
```

Penugasan No 4 - Eksperimen Selection Sort

```python
def selection_sort_max(data):
  n = len(data)
  for i in range(n - 1, 0, -1):
    max_index = 0
    for j in range(1, i + 1):
      if data[j] > data[max_index]:
        max_index = j
    data[i], data[max_index] = data[max_index], data[i]
  return data

angka = [64, 34, 25, 12, 22, 11, 90]
print("Data sebelum diurutkan: ", angka)
print("Data setelah Selection Sort (Maksimum): ", selection_sort_max(angka))
```

Penugasan No 5 - Eksperimen Insertion Sort

```python
def insertion_sort_nama(data):
  n = len(data)
  for i in range(1, n):
    key = data[i]
    j = i - 1

    while j >= 0 and data[j] > key:
      data[j + 1] = data[j]
      j -= 1
  return data

nama_mahasiswa = ["Zainab", "Budi", "Andi", "Siti", "Dyo", "Fajar"]
print("Data sebelum diurutkan: ", nama_mahasiswa)
print("Data setelah Insertion Sort (Alfabetis): ", insertion_sort_nama(nama_mahasiswa))
```

Penugasan No 6 - Studi Kasus Tambahan

```python
%%writefile nilai.py
def insertion_sort_ascending(data):
    n = len(data)
    for i in range(1, n):
        key = data[i]
        j = i - 1
        while j >= 0 and data[j][1] > key[1]:
            data[j + 1] = data[j]
            j -= 1

        data[j + 1] = key

    return data
```

Penugasan No 7 - Analisis Kompleksitas

```python
%%writefile bubble.py
def bubble_sort_count(data):
  n = len(data)
  count = 0
  for i in range(n):
    for j in range(0, n - i - 1):
      count += 1
      if data[j] > data[j + 1]:
        data[j], data[j + 1] = data[j + 1], data[j]
  return data, count
```

```python
%%writefile selection.py
def selection_sort_count(data):
  n = len(data)
  count = 0
  for i in range(n):
    min_index = i
    for j in range(i + 1, n):
      count += 1
      if data[j] < data[min_index]:
        min_index = j

    data[i], data[min_index] = data[min_index], data[i]

  return data, count
```

```python
%%writefile insertion.py
def insertion_sort_count(data):
  n = len(data)
  count = 0
  for i in range(1, n):
    key = data[i]
    j = i - 1

    while j >= 0 and data[j] > key:
      count += 1
      data[j + 1] = data[j]
      j -= 1
    if j >= 0:
      count += 1
    data[j + 1] = key
  return data, count
```

```python
import bubble
import selection
import insertion
import random

ukuran_data = [10, 50, 100]

print("=== HASIL ANALISIS JUMLAH PERBANDINGAN ===")

for n in ukuran_data:
    data_awal = random.sample(range(1, 1001), n)

    print(f"\nUkuran Input (N): {n}")

    _, count_bubble = bubble.bubble_sort_count(data_awal.copy())
    print(f"Bubble Sort   : {count_bubble} kali perbandingan")

    _, count_selection = selection.selection_sort_count(data_awal.copy())
    print(f"Selection Sort: {count_selection} kali perbandingan")

    _, count_insertion = insertion.insertion_sort_count(data_awal.copy())
    print(f"Insertion Sort: {count_insertion} kali perbandingan")

    print("-" * 30)
```

Penugasan No 8 - Proyek Mini

```python
def sort_barang(data, key_index, is_reverse=False):
  n = len(data)
  data_sorted = data.copy()

  for i in range(n):
    for j in range(0, n - i - 1):
      if is_reverse:
        if data_sorted[j][key_index] < data_sorted[j + 1][key_index]:
            data_sorted[j], data_sorted[j + 1] = data_sorted[j + 1], data_sorted[j]
      else:
        if data_sorted[j][key_index] > data_sorted[j + 1][key_index]:
            data_sorted[j], data_sorted[j + 1] = data_sorted[j + 1], data_sorted[j]
  return data_sorted

def tampilkan_tabel(judul, data):
  print(f"{judul}")
  if not data:
    print("(Data Barang Kosong)")
  else:
    print(f"{'No': <4} | {'Nama Barang' : <15} | {'Harga (Rp)' : <15}")
    for i, barang in enumerate(data):
      print(f"{i + 1 : <4} | {barang[0] : <15} | {barang[1] : <15}")

def main():
  daftar_barang = []
  while True:
    try:
      jumlah = int(input("Masukkan jumlah barang yang ingin diinput: "))
      if jumlah > 0:
        break
      print("Masukkan angka lebih dari 0.")
    except ValueError:
      print("Error : Masukkan angka yang valid!")

  print("Masukkan data barang: ")
  for i in range(jumlah):
    print(f"Data ke-{i+1}: ")
    nama = input("Nama barang: ")
    while True:
      try:
        harga = int(input("Harga barang:"))
        break
      except ValueError:
        print("Error: Harga harus angka!")
    daftar_barang.append((nama, harga))

  while True:
    print("\n=== MENU UTAMA ===")
    print("1. Tampilkan Barang")
    print("2. Tambah Barang Lagi")
    print("3. Urutkan berdasarkan NAMA (A-Z)")
    print("4. Urutkan berdasarkan NAMA (Z-A)")
    print("5. Urutkan berdasarkan HARGA (Termurah)")
    print("6. Urutkan berdasarkan HARGA (Termahal)")
    print("7. Keluar")

    pilihan = input("PIlih menu (1-7): ")

    if pilihan == '1':
      tampilkan_tabel("Daftar barang saat ini", daftar_barang)
    elif pilihan == '2':
      print("Tambah barang baru: ")
      nama = input("Nama barang: ")
      try:
        harga = int(input("Masukkkan Harga Barang: "))
        daftar_barang.append((nama, harga))
      except ValueError:
        print("Error: Harga harus angka!")
    elif pilihan == '3':
      hasil = sort_barang(daftar_barang, 0, False)
      tampilkan_tabel("Urut nama (A-Z)", hasil)
    elif pilihan == '4':
      hasil= sort_barang(daftar_barang, 0, True)
      tampilkan_tabel("Urut nama (Z-A)", hasil)
    elif pilihan == '5':
      hasil = sort_barang(daftar_barang, 1, False)
      tampilkan_tabel("Urut harga (termurah)", hasil)
    elif pilihan == '6':
      hasil = sort_barang(daftar_barang, 1, True)
      tampilkan_tabel("Urut harga (termahal)", hasil)
    elif pilihan == '7':
      print("Terimakasih, Sampai Jumpa!")
      break
    else:
      print("Pilihan tidak valid.")

if __name__ == "__main__":
  main()
```
